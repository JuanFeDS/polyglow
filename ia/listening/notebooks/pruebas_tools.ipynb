```python
def cortar_audio(ruta_entrada, ruta_salida, inicio_ms, fin_ms):
    """
    Corta un archivo de audio mp3 entre los milisegundos especificados.

    Parámetros:
        ruta_entrada (str): Ruta del archivo de audio de entrada.
        ruta_salida (str): Ruta donde se guardará el audio recortado.
        inicio_ms (int): Tiempo de inicio en milisegundos.
        fin_ms (int): Tiempo de fin en milisegundos.
    """
    # Cargar el archivo de audio
    audio = AudioSegment.from_mp3(ruta_entrada)
    
    # Cortar el fragmento
    fragmento = audio[inicio_ms:fin_ms]
    
    # Exportar el resultado
    fragmento.export(ruta_salida, format="mp3")
```

```python
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()  # asegúrate de tener OPENAI_API_KEY en .env

client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

audio_file = 'salida.mp3'  # ajusta la ruta a tu archivo

def transcribe(audio_file):
    with open(audio_file, "rb") as f:
        transcription = client.audio.transcriptions.create(
            model="whisper-1",
            file=f,
            # fuerza idioma, prueba con "en" o "es" según el audio real
            language="en",
            response_format="verbose_json"  # para ver segmentos completos
        )

    text = transcription.text
    return text
```

```python
from pathlib import Path
from openai import OpenAI

# Cliente nativo de OpenAI
client = OpenAI()

# Texto a convertir en voz
texto = "Hello, this is a test using OpenAI with LangChain to generate audio."

# Archivo de salida
output_file = Path("salida.mp3")

# Generación de audio con el modelo TTS
with client.audio.speech.with_streaming_response.create(
    model="gpt-4o-mini-tts",
    voice="alloy",  # voces disponibles: alloy, verse, coral, etc.
    input=texto,
) as response:
    response.stream_to_file(output_file)

print(f"✅ Audio generado en {output_file}")
```
